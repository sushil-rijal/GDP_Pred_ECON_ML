## Context & methods

### Key assumptions

- The model is dynamic: current log GDP per capita depends on its lag, private capital per capita, government capital per capita, human capital, and year effects.
- Difference GMM uses changes in variables and lagged levels as instruments. System GMM adds level equations instrumented by lagged first differences.
- GDP per capita, private capital, and government capital are endogenous. Effective labor per capita is first treated as predetermined and then as endogenous.
- Instruments are collapsed and restricted to lags 2?4 for endogenous variables. Predetermined effective labor uses lags 1?3 in the difference equation.
- Full year effects are retained; no separate linear trend is included.
- System GMM additionally requires a mean-stationarity restriction for its level moments.

The key validity requirement is absence of second-order serial correlation in differenced errors. A non-rejected Hansen test alone does not validate a model that fails this serial-correlation diagnostic.

## Results

The annual difference and system designs use 6,795 difference equations from 135 countries, with 71 and 76 collapsed instruments respectively?below the number of countries. Hansen J tests are not rejected. However, the AR(2) residual-correlation screens are significant in every version (p ? 0.002). This indicates that the annual moment conditions need further work before causal interpretation.

The persistence coefficient is near one in these annual specifications. That is another reason to avoid over-interpreting the short-run capital coefficients until the serial-correlation issue and the long-run time-series properties are addressed.


## Takeaways

- The GMM work uses differences for the dynamic equation and instruments potentially endogenous variables with lagged information.
- The restricted instrument count avoids the common problem of hundreds of instruments in a 135-country panel.
- The current annual specifications fail the key AR(2) diagnostic screen, so they are not the final causal model.
- Next options are to examine a more parsimonious lag structure, use non-overlapping multi-year periods, or adopt a dynamic-panel approach that explicitly accommodates remaining serial correlation. Any revision must again report its instrument count, Hansen test, and AR(1)/AR(2) diagnostics.

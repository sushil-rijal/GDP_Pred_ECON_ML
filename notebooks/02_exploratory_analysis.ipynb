## tl;dr

The analysis outcome is the natural log of observed PPP GDP per capita. The current construction yields 8,528 country-year observations across 163 countries. This notebook describes the realized analytical sample before econometric models are specified.

## Context & methods

PPP GDP per capita equals observed `GDP_rppp` (billions of PPP dollars) Ãƒâ€” 1,000,000,000 divided by observed total population. The outcome is then log-transformed. A country-year enters the outcome sample only when both values are observed and positive; no interpolation or statistical imputation is used.

```python
### 1. Load the audited panel
import pandas as pd
from src.descriptive import correlation_matrix, distribution_moments

panel = pd.read_csv("data/processed/analysis_panel.csv")
outcome_sample = panel.loc[panel["log_gdp_per_capita_ppp"].notna()].copy()
print(f"Outcome sample: {len(outcome_sample):,} country-years; {outcome_sample.country_code.nunique():,} countries")
```

```python
### 2. Describe the outcome and economically selected covariates
variables = [
    "log_gdp_per_capita_ppp", "gdp_per_capita_ppp", "ppp_capital",
    "priv_capital", "gov_capital", "labor_force", "human_capital_index",
    "tfp_index", "researchers_per_million", "energy_use_per_capita",
    "population_total", "age_dependency_ratio", "urban_population_pct",
]
available = [variable for variable in variables if variable in outcome_sample.columns]
descriptive_table = distribution_moments(outcome_sample, available)
descriptive_table.to_csv("results/tables/descriptive_statistics.csv", index=False)
descriptive_table
```

```python
### 3. Examine correlations before constructing feature blocks
correlation_variables = [
    "log_gdp_per_capita_ppp", "ppp_capital", "priv_capital", "gov_capital",
    "labor_force", "human_capital_index", "tfp_index", "researchers_per_million",
    "energy_use_per_capita", "age_dependency_ratio", "urban_population_pct",
]
correlation_variables = [variable for variable in correlation_variables if variable in outcome_sample.columns]
pearson_correlation = correlation_matrix(outcome_sample, correlation_variables, method="pearson")
spearman_correlation = correlation_matrix(outcome_sample, correlation_variables, method="spearman")
pearson_correlation.to_csv("results/tables/pearson_correlation.csv")
spearman_correlation.to_csv("results/tables/spearman_correlation.csv")
pearson_correlation.round(3)
```
